```python
# 1. for loop - basic iteration
for i in range(1, 6):
    print(i, i ** 2)
```

```python
# 2. while loop
n = 5
while n > 0:
    print(n)
    n -= 1
```

```python
# 3. for loop over a list of numerical values
marks = [45, 67, 89, 34, 92, 58]
for m in marks:
    result = "Pass" if m >= 40 else "Fail"
    print(m, "->", result)
```

```python
# 4. for loop over a dictionary
prices = {"apple": 40, "banana": 20, "mango": 80}
for item, price in prices.items():
    print(item, price)
```

```python
# 5. range() with step
for i in range(0, 20, 5):
    print(i)
```

```python
# 6. accumulator pattern - sum
sales = [1200, 1500, 900, 2100, 1750]
total = 0
for s in sales:
    total += s
print("Total:", total)
```

```python
# 7. accumulator pattern - average
average = total / len(sales)
print("Average:", average)
```

```python
# 8. while loop with a condition check
count = 0
i = 0
while i < len(marks):
    if marks[i] >= 50:
        count += 1
    i += 1
print("Students with 50+:", count)
```

```python
# 9. nested for loop
matrix = [[1, 2], [3, 4], [5, 6]]
for row in matrix:
    for val in row:
        print(val, end=" ")
print()
```

```python
# 10. min/max using a loop (no built-in min/max)
minimum = marks[0]
maximum = marks[0]
for m in marks:
    if m < minimum:
        minimum = m
    if m > maximum:
        maximum = m
print("Min:", minimum, "Max:", maximum)
```

## Program that processes a list of numerical values

```python
def classify_sales(values, target):
    above, below = [], []
    for v in values:
        if v >= target:
            above.append(v)
        else:
            below.append(v)
    return above, below

above, below = classify_sales(sales, 1500)
print("Above target:", above)
print("Below target:", below)
```

## Program that calculates summary statistics using loops

```python
def summary_stats(nums):
    total = 0
    minimum = nums[0]
    maximum = nums[0]
    for n in nums:
        total += n
        if n < minimum:
            minimum = n
        if n > maximum:
            maximum = n
    return {
        "sum": total,
        "average": total / len(nums),
        "min": minimum,
        "max": maximum,
    }

print(summary_stats(sales))
print(summary_stats(marks))
```

## Interview questions

**What is the difference between a for loop and a while loop?**
A `for` loop iterates over a known sequence (a list, range, dictionary) a fixed number of times. A `while` loop runs as long as a condition stays True, which is better when the number of iterations isn't known ahead of time.

**What does range() do?**
`range(start, stop, step)` generates a sequence of numbers without storing them all in memory. `range(5)` gives 0-4, `range(0,20,5)` gives 0,5,10,15 — stop is never included.

**Why are vectorized operations generally preferred over loops in data science?**
Loops process one element at a time in Python, with interpreter overhead on every iteration. Vectorized operations (pandas/NumPy) run as compiled C code across the whole array at once, which is dramatically faster on large datasets and also more concise to write.
